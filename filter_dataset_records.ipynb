```python
import pandas as pd

# 1. Dataset load
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Total records:", len(df))
print(df.columns.tolist())
df.head()

# 2. Example 1: Single condition
# Condition: Sex == 'female' -> Boolean mask create aagum, True rows mattum select aagum
females = df[df['Sex'] == 'female']
print("Female passengers:", len(females))
females.head()

# 3. Example 2: AND (&) - rendu condition um True irukkanum
# 1st class + survived
first_survived = df[(df['Pclass'] == 1) & (df['Survived'] == 1)]
print("1st class survivors:", len(first_survived))
first_survived.head()

# 4. Example 3: OR (|) - edhaavadhu oru condition True aana pothum
# Age > 60 OR Fare > 200
senior_or_rich = df[(df['Age'] > 60) | (df['Fare'] > 200)]
print("Age>60 or Fare>200:", len(senior_or_rich))
senior_or_rich.head()

# 5. Example 4: NOT (~) - condition ah reverse pannum
# Male illaadha passengers
not_male = df[~(df['Sex'] == 'male')]
print("Not male:", len(not_male))

# 6. Example 5: isin() - list la irukra values match pannum
class_1_2 = df[df['Pclass'].isin([1, 2])]
print("Class 1 or 2:", len(class_1_2))

# 7. Example 6: query() method
children_survived = df.query("Age < 18 and Survived == 1")
print("Children who survived:", len(children_survived))
children_survived.head()

# 8. Example 7: 3 conditions combine
# Female + 1st class + survived
combo = df[(df['Sex'] == 'female') & (df['Pclass'] == 1) & (df['Survived'] == 1)]
print("Female, 1st class, survived:", len(combo))
combo.head()

# 9. Summary
print("\n--- Summary ---")
print("Total:", len(df))
print("Females:", len(females))
print("1st class survivors:", len(first_survived))
print("Age>60 or Fare>200:", len(senior_or_rich))
print("Not male:", len(not_male))
print("Class 1 or 2:", len(class_1_2))
print("Children survived:", len(children_survived))
print("Female + 1st + survived:", len(combo))
```
